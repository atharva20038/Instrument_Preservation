```python
import os 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import json

# def find_file_meta()

with open('file_metadata.json', 'r') as f:
    metadata = json.load(f)
    
for key, value in metadata.items():
    filename = value.get('filename', 'N/A').split('-')[0] + ".wav"
    
    sampling_rate = value.get('sampling_rate', 'N/A')
    model_ = value.get('model', 'N/A')
    instrument = value.get('instrument', 'N/A')
    
    ## Find the filename 
    
    
    if sampling_rate == "16k" : 
        if model_ != "mustango" and model_ != "musicgen" : 
            proposed_path = os.path.join("../data_16k", "segmented", model_, instrument, filename)
            if os.path.exists(proposed_path):
                print(f"File found: {proposed_path}")
                
        else : 
            proposed_path = os.path.join(f"../../{model_}", "data", instrument, filename)
            if os.path.exists(proposed_path):
                print(f"File found: {proposed_path}")
                
    else : 
        if model_ != "mustango" and model_ != "musicgen" :
            proposed_path = os.path.join("../data", "segmented", instrument, filename)
            if os.path.exists(proposed_path):
                print(f"File found: {proposed_path}")
        else :
            proposed_path = os.path.join(f"../../{model_}", "data", model_, instrument, filename)
            if os.path.exists(proposed_path):
                print(f"File found: {proposed_path}")
                
    # print(proposed_path)
    abs_path = proposed_path
    
    value['abs_path'] = abs_path
    
    # print(f"{key}: {value}")
    # break
```

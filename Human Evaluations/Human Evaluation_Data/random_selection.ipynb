```python
import pandas as pd
import numpy as np
import os
```

```python
instrument = "Violin"
model = "real"

data_dir = f"data_16k/segmented/{model}/{instrument}"

files = os.listdir(data_dir)

song_title_list = [file.split("_")[1] for file in files]
song_title_list = list(set(song_title_list))

random_selection = np.random.choice(song_title_list, size=5, replace=False)

## For each file where song_title matches pick any clip at random

random_clips = []
for song_title in random_selection:
    matching_files = [file for file in files if song_title in file]
    random_clip = np.random.choice(matching_files)
    random_clips.append(random_clip)
    
print(random_clips)
```
